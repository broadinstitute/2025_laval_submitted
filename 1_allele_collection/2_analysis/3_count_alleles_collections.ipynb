# Count alleles and variants in the different collections

```python
import polars as pl
import pandas as pd
```

```python
inputs_dir = "../3_outputs"
```

```python
metadata = pl.DataFrame(pd.read_csv(f"{inputs_dir}/varchamp_consolidated_alleles_clinvar_gnomad_sequence_confirmed.tsv", sep="\t"))
collections = metadata.select("collection").to_series().unique().to_list()
```

```python
for coll in collections:
    coll_df = metadata.filter(pl.col("collection") == coll)

    genes = coll_df.select("symbol").to_series().unique().to_list()
    alleles = coll_df.select(["symbol", "aa_change"]).unique()

    print(f"For collection {coll}, there are {len(genes)} genes and {alleles.shape[0]} alleles.")

all_genes = metadata.select("symbol").unique().shape[0]
all_alleles = metadata.select(["symbol", "aa_change"]).unique().shape[0]

print(f"Overall, there are {all_genes} genes and {all_alleles} alleles.")
```
