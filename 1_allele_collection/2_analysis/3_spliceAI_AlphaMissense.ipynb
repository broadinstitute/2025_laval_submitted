```python
import pandas as pd
import polars as pl
```

```python
inputs_dir = "../1_inputs"
outputs_dir = "../3_outputs"
```

# SpliceAI

```python
splice_df = pd.read_csv(f"{inputs_dir}/ccsb_mutations_with_spliceai_scores.tsv", sep="\t", low_memory=False)
splice_df
```

```python
allele_col = pl.read_parquet(f"{outputs_dir}/VarChAMP_Allele_Collection_1Perc_ClinVar_dbNSFP_pLDDT_RSA_G2P_OMIM_MOI.parquet").to_pandas()
allele_col
```

```python
# Create a mapping DataFrame with just the columns we need
splice_mapping = splice_df[['orf_id', 'nt_change', 'any_DS_above_0.2']].rename(
    columns={'any_DS_above_0.2': 'splicing_variant'}
)

# Merge onto allele_col
allele_col = allele_col.merge(
    splice_mapping,
    on=['orf_id', 'nt_change'],
    how='left'
)

# Fill NaN values with False (variants not found in splice_df)
allele_col['splicing_variant'] = allele_col['splicing_variant'].fillna(False)

print(f"Total variants in allele_col: {len(allele_col)}")
print(f"Splicing variants (any_DS > 0.2): {allele_col['splicing_variant'].sum()}")

# allele_col.drop(columns=["seq", "seq_aa"]).to_csv(f"{outputs_dir}/VarChAMP_Allele_Collection_1Perc_ClinVar_dbNSFP_pLDDT_RSA_G2P_OMIM_MOI_SpliceAI.tsv", sep="\t", index=False)
allele_col
```

# AlphaMissense

```python
# TODO: add zenodo download to download_data.sh

AM_df = pd.read_csv(f'{inputs_dir}/raw_inputs/AlphaMissense/AlphaMissense_hg38.tsv', sep='\t', skiprows=3)
AM_df
```

```python
# Fast mapping using merge instead of dictionary lookup for allele_col

# Prepare allele_col columns
allele_col['nuc_loc'] = pd.to_numeric(allele_col['nuc_loc'], errors='coerce').astype('Int64')

# Get unique variants from allele_col to filter AM_df first
unique_positions = allele_col['nuc_loc'].dropna().unique()

# Filter AM_df to only positions we care about (MUCH faster)
AM_filtered = AM_df[AM_df['POS'].isin(unique_positions)].copy()

# Convert #CHROM: chr1 -> 1, chrX -> 23, chrY -> 24
def parse_chrom(chrom_str):
    chrom = str(chrom_str).replace('chr', '')
    if chrom == 'X':
        return 23
    elif chrom == 'Y':
        return 24
    elif chrom == 'M' or chrom == 'MT':
        return 25
    else:
        try:
            return int(chrom)
        except ValueError:
            return None

AM_filtered['chr_num'] = AM_filtered['#CHROM'].apply(parse_chrom).astype('Int64')

# Drop duplicates BEFORE merge
AM_filtered_dedup = AM_filtered.drop_duplicates(subset=['chr_num', 'POS', 'REF', 'ALT'], keep='first')

print(f"Filtered AM_df from {len(AM_df):,} to {len(AM_filtered):,} rows")
print(f"After dedup: {len(AM_filtered_dedup):,} rows")

# Now merge on chromosome + position + alleles
merged = allele_col.merge(
    AM_filtered_dedup[['chr_num', 'POS', 'REF', 'ALT', 'am_pathogenicity']],
    left_on=['chr_num', 'nuc_loc', 'ref_allele', 'alt_allele'],
    right_on=['chr_num', 'POS', 'REF', 'ALT'],
    how='left',
    suffixes=('', '_am')
)

# Verify shapes match
print(f"allele_col shape: {allele_col.shape}")
print(f"merged shape: {merged.shape}")

# Copy the score to allele_col
allele_col['am_pathogenicity'] = merged['am_pathogenicity'].values

# Check results
print(f"Total variants in allele_col: {len(allele_col)}")
print(f"Variants with AM score mapped: {allele_col['am_pathogenicity'].notna().sum()}")
print(f"Variants without AM score: {allele_col['am_pathogenicity'].isna().sum()}")
```

# Met1Xxx removal

```python
# Remove Met1Xxx variants as we cannot assay those variants

if 'aa_change' in allele_col.columns:
    allele_col = allele_col[
        allele_col['aa_change'].isnull() |
        ~allele_col['aa_change'].str.contains(r'^Met1[A-Z][a-z]{2}$', na=False)
    ]
```

```python
# Save final allele collection with SpliceAI and AlphaMissense scores
allele_col.to_csv(f'{outputs_dir}/VarChAMP_Allele_Collection_1Perc_ClinVar_dbNSFP_pLDDT_RSA_G2P_OMIM_MOI_SpliceAI_AM.tsv', sep='\t', index=False)

# Save a subset for downstream analysis
analysis_df = allele_col[allele_col['am_pathogenicity'].notna()]
analysis_df.to_csv(f'{outputs_dir}/VarChAMP_Allele_Collection_1Perc_ClinVar_dbNSFP_pLDDT_RSA_G2P_OMIM_MOI_SpliceAI_AM_Analysis.tsv', sep='\t', index=False)
```
