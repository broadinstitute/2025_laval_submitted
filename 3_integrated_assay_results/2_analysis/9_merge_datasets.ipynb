```python
# Loading allele collection dataset

allele_col = pd.read_csv('../../1_allele_collection/3_outputs/VarChAMP_Allele_Collection_1Perc_ClinVar_dbNSFP_pLDDT_RSA_G2P_OMIM_MOI_SpliceAI_AM.tsv', sep='\t')
allele_col
```

```python
# Check if all Sahni variants are in the allele collection
missing_in_allele_col = sahni_vars[~sahni_vars.ccsb_mutation_id.isin(allele_col.ccsb_mutation_id)]
print(f"Number of Sahni variants missing in allele collection: {len(missing_in_allele_col)}")
missing_in_allele_col
```

```python
# Mapping ClinVar clinical significance from allele_col (clinvar_clnsig_clean) to Sahni variants using ccsb_mutation_id

allele_col_dedup = allele_col.drop_duplicates(subset=['gene_variant'])

sahni_vars = sahni_vars.merge(
    allele_col_dedup[['ccsb_mutation_id', 'symbol', 'clinvar_clnsig_clean']],
    on='ccsb_mutation_id',
    how='left',
    suffixes=('', '_allele_col')
)
# sahni_vars.drop_duplicates(subset=['mutation_id_rc4'], inplace=True)
sahni_vars['symbol'] = sahni_vars['symbol_allele_col']
sahni_vars.drop(columns=['symbol_allele_col'], inplace=True)
sahni_vars['gene_variant'] = sahni_vars['symbol'] + '_' + sahni_vars['aa_change']
sahni_vars = sahni_vars[sahni_vars.gene_variant.notna()]
sahni_vars
```

```python
# Compare clinical_significance and clinvar_clnsig_clean in sahni_vars
print("=== clinical_significance unique values ===")
print(sahni_vars['clinical_significance'].value_counts(dropna=False))

print("\n=== clinvar_clnsig_clean unique values ===")
print(sahni_vars['clinvar_clnsig_clean'].value_counts(dropna=False))

# Cross-tab
both_have = sahni_vars[
    sahni_vars['clinical_significance'].notna() & 
    sahni_vars['clinvar_clnsig_clean'].notna()
].copy()

print(f"\nVariants with both annotations: {len(both_have)}")

if len(both_have) > 0:
    ct = pd.crosstab(
        both_have['clinical_significance'], 
        both_have['clinvar_clnsig_clean'],
        margins=True
    )
    print("\n--- Cross-tab: clinical_significance (rows) vs clinvar_clnsig_clean (columns) ---")
    print(ct.to_string())
    
    concordant = (both_have['clinical_significance'] == both_have['clinvar_clnsig_clean']).sum()
    discordant = (both_have['clinical_significance'] != both_have['clinvar_clnsig_clean']).sum()
    print(f"\n  Concordant: {concordant} ({100*concordant/len(both_have):.1f}%)")
    print(f"  Discordant: {discordant} ({100*discordant/len(both_have):.1f}%)")
    
    if discordant > 0:
        disc = both_have[both_have['clinical_significance'] != both_have['clinvar_clnsig_clean']]
        print(f"\n--- Discordant transitions ---")
        print(pd.crosstab(disc['clinical_significance'], disc['clinvar_clnsig_clean']).to_string())
        print(f"\n--- Example discordant variants ---")
        print(disc[['gene_variant', 'clinical_significance', 'clinvar_clnsig_clean']].head(20).to_string(index=False))

# Coverage comparison
only_clinsig = sahni_vars[
    sahni_vars['clinical_significance'].notna() & 
    sahni_vars['clinvar_clnsig_clean'].isna()
]
only_clean = sahni_vars[
    sahni_vars['clinical_significance'].isna() & 
    sahni_vars['clinvar_clnsig_clean'].notna()
]
print(f"\n  Only clinical_significance:  {len(only_clinsig)}")
print(f"  Only clinvar_clnsig_clean:   {len(only_clean)}")
print(f"  Neither:                     {sahni_vars['clinical_significance'].isna().sum() - len(only_clean)}")
```

```python
rol_df = pd.read_excel(f'{datasets_dir}/Rolland/1-s2.0-S0092867414014226-mmc4.xlsx')
rol_df.rename({'aa-change':'aa_change'}, axis=1 ,inplace=True)

# No Clinvar mut? what does disease mutation/common variant means? gnomad frequent/rare?

from Bio.SeqUtils import seq3

rol_df['aa_change'] = [f'{seq3(aa_change[0])}{aa_change[1:-1]}{seq3(aa_change[-1])}' for aa_change in rol_df.aa_change]
rol_df.columns
```

```python
cols_to_keep = ['category', 'gene_symbol', 'aa_change',
       'Interactor_gene_symbol', 'Y2H_score',
       'Phenotype']

rol_df = rol_df[cols_to_keep]

# Update symbol ADCK3 to COQ8A to match allele collection
rol_df['gene_symbol'] = rol_df['gene_symbol'].replace({'ADCK3': 'COQ8A'})

rol_df['gene_variant'] = rol_df.gene_symbol + '_' + rol_df.aa_change
rol_df
```

```python
def edgotype_status(score_list):
    score_list = score_list.to_list()
    if len(score_list) == 1:
        if pd.isnull(score_list[0]):
            return np.nan
        elif score_list[0] == 1:
            return 'Unperturbed-degree-1'
        elif score_list[0] == 0:
            return 'Perturbed-degree-1'
        else:
            return np.nan
    if pd.isnull(score_list).all():
        return np.nan
    if 0 in score_list and 1 in score_list:
        return 'Edgetic'
    elif score_list.count(0) >= 2:
        return 'Quasi-null'
    elif score_list.count(1) >= 2:
        return 'Quasi-wild-type'
    else:
        return np.nan

# Group by variant (gene + aa_change) and assign edgotype
rol_df['Edgotype_class'] = rol_df.groupby('gene_variant')['Y2H_score'].transform(edgotype_status)

rol_df.drop_duplicates(subset=['gene_variant'], inplace=True)
rol_df.drop(columns=['Interactor_gene_symbol', 'Y2H_score'], inplace=True)
rol_df
    
```

## Check Overlap of variants across datasets

```python
# Check for overlapping gene_variants across edgotyping datasets
edgotype_datasets = {
    'Sahni': set(sahni_vars[sahni_vars.Edgotype_class.notna()].gene_variant),
    'Rolland': set(rol_df[rol_df.Edgotype_class.notna()].gene_variant),
    'Luck': set(luck_df[luck_df.Edgotype_class.notna()].gene_variant),
    'Fragoza_ExAC': set(frag_exac_df[frag_exac_df.Edgotype_class.notna()].gene_variant),
    'Fragoza_COSMIC': set(frag_cosmic[frag_cosmic.Edgotype_class.notna()].gene_variant),
    'Fragoza_HGMD': set(frag_hgmd[frag_hgmd.Edgotype_class.notna()].gene_variant),
}

from itertools import combinations

for (name_a, set_a), (name_b, set_b) in combinations(edgotype_datasets.items(), 2):
    overlap = set_a & set_b
    if overlap:
        print(f"{name_a} ∩ {name_b}: {len(overlap)} overlapping variants")
        print(f"  Examples: {list(overlap)[:5]}")
    else:
        print(f"{name_a} ∩ {name_b}: no overlap")
```

```python
# For overlapping variants, compare their Edgotype_class across datasets
edgotype_dfs = {
    'Sahni': sahni_vars[sahni_vars.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
    'Rolland': rol_df[rol_df.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
    'Luck': luck_df[luck_df.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
    'Fragoza_ExAC': frag_exac_df[frag_exac_df.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
    'Fragoza_COSMIC': frag_cosmic[frag_cosmic.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
    'Fragoza_HGMD': frag_hgmd[frag_hgmd.Edgotype_class.notna()][['gene_variant', 'Edgotype_class']],
}

from itertools import combinations

for (name_a, df_a), (name_b, df_b) in combinations(edgotype_dfs.items(), 2):
    merged = df_a.merge(df_b, on='gene_variant', suffixes=(f'_{name_a}', f'_{name_b}'))
    if len(merged) == 0:
        continue
    
    col_a = f'Edgotype_class_{name_a}'
    col_b = f'Edgotype_class_{name_b}'
    
    concordant = (merged[col_a] == merged[col_b]).sum()
    discordant = (merged[col_a] != merged[col_b]).sum()
    
    print(f"\n{name_a} ∩ {name_b}: {len(merged)} overlapping variants")
    print(f"  Concordant: {concordant}, Discordant: {discordant}")
    
    if discordant > 0:
        disc_df = merged[merged[col_a] != merged[col_b]]
        print(f"  Discordant cases:")
        print(disc_df[['gene_variant', col_a, col_b]].to_string(index=False))
```

## Merging step

```python
# Build unified Edgotype_class with priority: Sahni > Rolland > Luck > Fragoza
# For Fragoza-only variants in multiple Fragoza sub-datasets: concordant -> keep, discordant -> discard

# Step 1: Identify Fragoza-only discordant variants to discard
sahni_rolland_luck = edgotype_datasets['Sahni'] | edgotype_datasets['Rolland'] | edgotype_datasets['Luck']
fragoza_only = {}
for name in ['Fragoza_ExAC', 'Fragoza_COSMIC', 'Fragoza_HGMD']:
    fragoza_only[name] = edgotype_datasets[name] - sahni_rolland_luck

discard_variants = set()
fragoza_names = ['Fragoza_ExAC', 'Fragoza_COSMIC', 'Fragoza_HGMD']
for (name_a, name_b) in combinations(fragoza_names, 2):
    overlap = fragoza_only[name_a] & fragoza_only[name_b]
    if not overlap:
        continue
    df_a = edgotype_dfs[name_a].set_index('gene_variant')
    df_b = edgotype_dfs[name_b].set_index('gene_variant')
    for v in overlap:
        if df_a.loc[v, 'Edgotype_class'] != df_b.loc[v, 'Edgotype_class']:
            discard_variants.add(v)

print(f"Discarding {len(discard_variants)} Fragoza-only discordant variants:")
for v in sorted(discard_variants):
    print(f"  {v}")

# Step 2: Build priority-based Edgotype_class lookup
# Priority order: Sahni, Rolland, Luck, Fragoza_ExAC, Fragoza_COSMIC, Fragoza_HGMD
priority_order = ['Sahni', 'Rolland', 'Luck', 'Fragoza_ExAC', 'Fragoza_COSMIC', 'Fragoza_HGMD']

edgotype_lookup = {}
# Go in reverse priority so higher priority overwrites
for name in reversed(priority_order):
    df = edgotype_dfs[name]
    for _, row in df.iterrows():
        v = row['gene_variant']
        if v in discard_variants and name.startswith('Fragoza'):
            continue
        edgotype_lookup[v] = row['Edgotype_class']

print(f"\nTotal variants with Edgotype_class: {len(edgotype_lookup)}")

# Step 3: Also track which dataset the edgotype came from
edgotype_source = {}
for name in reversed(priority_order):
    df = edgotype_dfs[name]
    for _, row in df.iterrows():
        v = row['gene_variant']
        if v in discard_variants and name.startswith('Fragoza'):
            continue
        edgotype_source[v] = name

# Step 4: Merge all datasets (outer join) keeping dataset-specific metadata
# but using a single Edgotype_class column

# --- Sahni (combined PPI + Chaperone) ---
sahni_combined = sahni_vars[['symbol', 'aa_change', 'gene_variant', 'clinical_significance', 'Category', 'Sahni_Chaperone_binding']].copy()
sahni_combined = sahni_combined.drop_duplicates('gene_variant')
sahni_combined.rename(columns={
    'symbol': 'gene_symbol',
    'clinical_significance': 'Sahni_clinical_significance',
    'Category': 'Sahni_Category',
}, inplace=True)


# --- Rolland metadata ---
rol_harmonized = rol_df[['gene_symbol', 'aa_change', 'gene_variant', 'category', 'Phenotype']].copy()
rol_harmonized.rename(columns={
    'category': 'Rolland_category',
    'Phenotype': 'Rolland_Phenotype',
}, inplace=True)

# --- Luck metadata ---
luck_harmonized = luck_df[['CG_symbol', 'mutation', 'gene_variant', 'assay_id', 'pathogenecity', 'omim_ids']].copy()
luck_harmonized.rename(columns={
    'CG_symbol': 'gene_symbol',
    'mutation': 'aa_change',
    'assay_id': 'Luck_assay_id',
    'pathogenecity': 'Luck_pathogenecity',
    'omim_ids': 'Luck_omim_ids',
}, inplace=True)

# --- Lacoste Imaging ---
lacoste_harmonized = lacoste_df[['Gene', 'aa_change', 'ClinVar class', 'Phenotype annotation',
                                  'dbSNP_ID', 'Uniprot', 'Mislocalized', 'Impact score']].copy()
lacoste_harmonized.rename(columns={
    'Gene': 'gene_symbol',
    'ClinVar class': 'Lacoste_ClinVar_class',
    'Phenotype annotation': 'Lacoste_Phenotype_annotation',
    'dbSNP_ID': 'Lacoste_dbSNP_ID',
    'Uniprot': 'Lacoste_Uniprot',
    'Mislocalized': 'Lacoste_Mislocalized',
    'Impact score': 'Lacoste_Impact_score',
}, inplace=True)
lacoste_harmonized['gene_variant'] = lacoste_harmonized.gene_symbol + '_' + lacoste_harmonized.aa_change

# --- Fragoza ExAC metadata ---
frag_exac_harmonized = frag_exac_df[['target_symbol', 'aa_change', 'gene_variant', 'dbSNP_id', 'Mutation', 'AF_all']].copy()
frag_exac_harmonized.rename(columns={
    'target_symbol': 'gene_symbol',
    'dbSNP_id': 'Fragoza_ExAC_dbSNP_id',
    'Mutation': 'Fragoza_ExAC_Mutation',
    'AF_all': 'Fragoza_ExAC_AF_all',
}, inplace=True)

# --- Fragoza COSMIC metadata ---
frag_cosmic_harmonized = frag_cosmic[['target_symbol', 'aa_change', 'gene_variant', 'dbSNP_id', 'Mutation']].copy()
frag_cosmic_harmonized.rename(columns={
    'target_symbol': 'gene_symbol',
    'dbSNP_id': 'Fragoza_COSMIC_dbSNP_id',
    'Mutation': 'Fragoza_COSMIC_Mutation',
}, inplace=True)

# --- Fragoza HGMD metadata ---
frag_hgmd_harmonized = frag_hgmd[['target_symbol', 'aa_change', 'gene_variant', 'Mutation']].copy()
frag_hgmd_harmonized.rename(columns={
    'target_symbol': 'gene_symbol',
    'Mutation': 'Fragoza_HGMD_Mutation',
}, inplace=True)

# --- Fragoza Abundance ---
frag_abun_harmonized = frag_abun[['target_symbol', 'aa_change', 'HGVS_cDNA', 'Status']].copy()
frag_abun_harmonized.rename(columns={
    'target_symbol': 'gene_symbol',
    'Status': 'Fragoza_Abundance_Status',
    'HGVS_cDNA': 'Fragoza_Abundance_HGVS_cDNA',
}, inplace=True)
frag_abun_harmonized['gene_variant'] = frag_abun_harmonized.gene_symbol + '_' + frag_abun_harmonized.aa_change

# --- Merge all on gene_variant (outer joins) ---
all_datasets = [
    sahni_combined,
    rol_harmonized,
    luck_harmonized,
    lacoste_harmonized,
    frag_exac_harmonized,
    frag_cosmic_harmonized,
    frag_hgmd_harmonized,
    frag_abun_harmonized,
]

from functools import reduce

merged_df = reduce(
    lambda left, right: left.merge(
        right.drop(columns=['gene_symbol', 'aa_change'], errors='ignore'),
        on='gene_variant',
        how='outer'
    ),
    all_datasets
)

# Fill gene_symbol and aa_change from gene_variant for rows that came from right-only joins
merged_df['gene_symbol'] = merged_df['gene_symbol'].fillna(merged_df['gene_variant'].str.rsplit('_', n=1).str[0])
merged_df['aa_change'] = merged_df['aa_change'].fillna(merged_df['gene_variant'].str.rsplit('_', n=1).str[1])

# Step 5: Add unified Edgotype_class and source columns
merged_df['Edgotype_class'] = merged_df['gene_variant'].map(edgotype_lookup)
merged_df['Edgotype_source'] = merged_df['gene_variant'].map(edgotype_source)

print(f"Merged table: {merged_df.shape}")
print(f"Unique variants: {merged_df.gene_variant.nunique()}")
print(f"Variants with Edgotype_class: {merged_df.Edgotype_class.notna().sum()}")
print(f"\nEdgotype source breakdown:")
print(merged_df.Edgotype_source.value_counts())
merged_df
```

```python
import re
from collections import defaultdict

# Parse aa_change into (ref_aa, position, alt_aa)
def parse_aa_change(aa_change):
    match = re.match(r'^([A-Z][a-z]{2})(\d+)([A-Z][a-z]{2})$', aa_change)
    if match:
        return match.group(1), int(match.group(2)), match.group(3)
    return None, None, None

# Group variants by gene
gene_variants = defaultdict(list)
for gv in full_df['gene_variant'].dropna().unique():
    parts = gv.rsplit('_', 1)
    if len(parts) == 2:
        gene, aa = parts
        ref, pos, alt = parse_aa_change(aa)
        if pos is not None:
            gene_variants[gene].append((ref, pos, alt, gv))

# For each gene, find pairs with same ref_aa and alt_aa but different positions
print("=== Potential isoform-offset duplicates ===\n")
isoform_duplicates = []

for gene, variants in sorted(gene_variants.items()):
    if len(variants) < 2:
        continue
    
    # Group by (ref_aa, alt_aa)
    aa_groups = defaultdict(list)
    for ref, pos, alt, gv in variants:
        aa_groups[(ref, alt)].append((pos, gv))
    
    for (ref, alt), positions in aa_groups.items():
        if len(positions) < 2:
            continue
        positions.sort()
        # Check all pairs for consistent offsets within this gene
        for i in range(len(positions)):
            for j in range(i+1, len(positions)):
                pos_i, gv_i = positions[i]
                pos_j, gv_j = positions[j]
                offset = pos_j - pos_i
                isoform_duplicates.append({
                    'gene': gene,
                    'variant_1': gv_i,
                    'variant_2': gv_j,
                    'offset': offset,
                    'ref_aa': ref,
                    'alt_aa': alt,
                })

dup_df = pd.DataFrame(isoform_duplicates)
print(f"Total potential isoform-offset duplicate pairs: {len(dup_df)}")

# Find genes with consistent offsets (multiple pairs with same offset)
if len(dup_df) > 0:
    consistent = dup_df.groupby(['gene', 'offset']).filter(lambda x: len(x) >= 2)
    consistent_summary = consistent.groupby(['gene', 'offset']).agg(
        n_pairs=('variant_1', 'count'),
        examples=('variant_1', lambda x: list(x)[:3])
    ).reset_index().sort_values('n_pairs', ascending=False)
    
    print(f"\nGenes with consistent offsets (≥2 pairs same offset):")
    print(f"{'Gene':<15} {'Offset':<8} {'Pairs':<6} Examples")
    print("-" * 80)
    for _, row in consistent_summary.iterrows():
        gene_rows = consistent[(consistent.gene == row.gene) & (consistent.offset == row.offset)]
        pairs_str = '; '.join([f"{r.variant_1} / {r.variant_2}" for _, r in gene_rows.head(3).iterrows()])
        print(f"{row.gene:<15} {row.offset:<8} {row.n_pairs:<6} {pairs_str}")
    
    print(f"\n⚠️  These {len(consistent_summary)} gene-offset groups likely represent the same variants ")
    print(f"   mapped to different transcript isoforms and should be deduplicated.")
    print(f"   Total variant pairs to resolve: {len(consistent)}")
```

```python
# === Merge cross-dataset isoform duplicates (Sahni aa_change is canonical) ===

full_bu = full_df.copy()  # backup before merging

# Step 1: Identify cross-dataset pairs only (different_datasets pattern)
cross_dataset_pairs = pair_src_df[pair_src_df.pattern == 'different_datasets'].copy()
print(f"Cross-dataset isoform duplicate pairs to merge: {len(cross_dataset_pairs)}")
print(f"\nSource combo breakdown:")
cross_dataset_pairs['source_combo'] = cross_dataset_pairs.apply(
    lambda r: ' ↔ '.join(sorted([str(r.source_1), str(r.source_2)])), axis=1
)
print(cross_dataset_pairs['source_combo'].value_counts().to_string())

# Step 2: For each pair, determine which variant to keep (Sahni-side) and which to fold in
# Priority: variant from 1pct/Sahni keeps its gene_variant; Lacoste/Fragoza/other folds in

source_priority = ['1pct', 'Sahni', 'Rolland', 'Luck', 'Fragoza', 'Lacoste']

def get_priority(source_str):
    """Lower number = higher priority (keep this variant's aa numbering)."""
    if pd.isna(source_str):
        return 999
    sources = source_str.split('; ')
    best = 999
    for s in sources:
        if s in source_priority:
            best = min(best, source_priority.index(s))
    return best

# Build merge plan: for each pair, keep_variant (canonical) absorbs drop_variant
merge_plan = []
for _, row in cross_dataset_pairs.iterrows():
    pri1 = get_priority(row['source_1'])
    pri2 = get_priority(row['source_2'])
    
    if pri1 <= pri2:
        keep_variant = row['variant_1']
        drop_variant = row['variant_2']
    else:
        keep_variant = row['variant_2']
        drop_variant = row['variant_1']
    
    merge_plan.append({
        'gene': row['gene'],
        'offset': row['offset'],
        'keep_variant': keep_variant,
        'drop_variant': drop_variant,
        'keep_source': row['source_1'] if pri1 <= pri2 else row['source_2'],
        'drop_source': row['source_2'] if pri1 <= pri2 else row['source_1'],
    })

merge_plan_df = pd.DataFrame(merge_plan)
print(f"\nMerge plan: {len(merge_plan_df)} pairs")
print(f"  Variants to drop (fold into canonical): {merge_plan_df['drop_variant'].nunique()}")
print(f"\nKeep → Drop direction:")
print(merge_plan_df.groupby(['keep_source', 'drop_source']).size().to_string())

# Step 3: For each drop_variant, fold its data columns into the keep_variant row
# Columns that come from Lacoste (the main "drop" source):
lacoste_cols = ['Lacoste_ClinVar_class', 'Lacoste_Phenotype_annotation', 
                'Lacoste_dbSNP_ID', 'Lacoste_Uniprot', 'Lacoste_Mislocalized', 
                'Lacoste_Impact_score']

# Build a mapping: drop_variant -> keep_variant
drop_to_keep = dict(zip(merge_plan_df['drop_variant'], merge_plan_df['keep_variant']))

# For each drop_variant, transfer non-NaN values to the keep_variant row
full_df_indexed = full_df.set_index('gene_variant')

transferred = 0
for drop_var, keep_var in drop_to_keep.items():
    if drop_var not in full_df_indexed.index or keep_var not in full_df_indexed.index:
        continue
    
    drop_row = full_df_indexed.loc[drop_var]
    
    # Transfer all columns where keep_variant is NaN but drop_variant has data
    for col in full_df_indexed.columns:
        if col in ['gene_symbol', 'aa_change', 'gene_variant', 'gene_symbol_old', 'alias_symbol']:
            continue  # Don't overwrite identity columns
        
        # Handle potential duplicate index (shouldn't happen after dedup, but be safe)
        keep_val = full_df_indexed.at[keep_var, col] if keep_var in full_df_indexed.index else np.nan
        drop_val = drop_row[col] if not isinstance(drop_row, pd.DataFrame) else drop_row.iloc[0][col]
        
        if pd.isna(keep_val) and pd.notna(drop_val):
            full_df_indexed.at[keep_var, col] = drop_val
            transferred += 1
    
    # Update Source columns for keep_variant
    keep_source = full_df_indexed.at[keep_var, 'Source']
    drop_source = drop_row['Source'] if not isinstance(drop_row, pd.DataFrame) else drop_row.iloc[0]['Source']
    
    if pd.notna(drop_source):
        if pd.notna(keep_source):
            all_sources = keep_source.split('; ') + drop_source.split('; ')
            unique_sources = list(dict.fromkeys(all_sources))
            full_df_indexed.at[keep_var, 'Source'] = '; '.join(unique_sources)
        else:
            full_df_indexed.at[keep_var, 'Source'] = drop_source

print(f"\nTransferred {transferred} non-NaN values from drop variants to keep variants")

# Step 4: Remove the drop_variants from the dataset
variants_to_drop = set(drop_to_keep.keys())
n_before = len(full_df_indexed)
full_df = full_df_indexed.drop(index=list(variants_to_drop & set(full_df_indexed.index)), errors='ignore').reset_index()
n_after = len(full_df)
print(f"\nRemoved {n_before - n_after} isoform-duplicate rows")

# Step 5: Recalculate Source_Simple and n_sources
full_df['Source_Simple'] = full_df['Source'].apply(get_sources_simple)
full_df['n_sources'] = full_df['Source'].apply(lambda x: len(x.split('; ')) if pd.notna(x) else 0)

print(f"\nFinal dataset after isoform deduplication: {full_df.shape}")
print(f"Unique variants: {full_df.gene_variant.nunique()}")

# Step 6: Show what was merged (for verification)
print(f"\n=== Merged isoform pairs (showing transferred Lacoste data) ===")
merged_variants = set(drop_to_keep.values())
verify_df = full_df[full_df.gene_variant.isin(merged_variants)]
verify_cols = ['gene_variant', 'Source_Simple'] + [c for c in lacoste_cols if c in full_df.columns]
print(verify_df[verify_cols].to_string(index=False))

# Step 7: Log the mapping for traceability
print(f"\n=== Isoform mapping log (drop → keep) ===")
print(f"{'Drop variant (alt isoform)':<35} {'Keep variant (canonical)':<35} {'Drop source':<15} {'Keep source':<15}")
print("-" * 105)
for _, row in merge_plan_df.iterrows():
    print(f"{row['drop_variant']:<35} {row['keep_variant']:<35} {row['drop_source']:<15} {row['keep_source']:<15}")
```

# Merging Abundance and Imaging data

```python
# === Unify Imaging and Abundance hits across datasets ===

# -------------------------------------------------------
# 1. IMAGING HIT UNIFIED
# -------------------------------------------------------

# Audit values
print("=== Imaging columns audit ===")
print(f"\nimaging_hit (1pct):")
print(full_df['imaging_hit'].value_counts(dropna=False))
print(f"\nLacoste_Mislocalized:")
print(full_df['Lacoste_Mislocalized'].value_counts(dropna=False))

# Harmonize Lacoste_Mislocalized (1.0/0.0 float) to True/False
full_df['_lacoste_imaging_hit'] = full_df['Lacoste_Mislocalized'].map({1.0: True, 0.0: False})

# Check concordance for overlapping variants
img_both = full_df[full_df['imaging_hit'].notna() & full_df['_lacoste_imaging_hit'].notna()].copy()
print(f"\n=== Imaging concordance (variants with both 1pct and Lacoste) ===")
print(f"Variants with both: {len(img_both)}")

if len(img_both) > 0:
    img_both['_img_concordant'] = img_both['imaging_hit'] == img_both['_lacoste_imaging_hit']
    concordant = img_both['_img_concordant'].sum()
    discordant = (~img_both['_img_concordant']).sum()
    print(f"  Concordant: {concordant} ({100*concordant/len(img_both):.1f}%)")
    print(f"  Discordant: {discordant} ({100*discordant/len(img_both):.1f}%)")
    
    if discordant > 0:
        ct = pd.crosstab(
            img_both['imaging_hit'].map({True: '1pct_True', False: '1pct_False'}),
            img_both['_lacoste_imaging_hit'].map({True: 'Lacoste_True', False: 'Lacoste_False'}),
            margins=True
        )
        print(f"\n  Cross-tab:")
        print(ct.to_string())
        
        disc = img_both[~img_both['_img_concordant']]
        print(f"\n  Discordant variants:")
        print(disc[['gene_variant', 'imaging_hit', 'Lacoste_Mislocalized', 'Source_Simple']].to_string(index=False))

# Build unified column: 1pct > Lacoste
full_df['imaging_hit_unified'] = full_df['imaging_hit'].where(
    full_df['imaging_hit'].notna(),
    full_df['_lacoste_imaging_hit']
)

# Track source
def get_imaging_source(row):
    if pd.notna(row['imaging_hit']):
        return '1pct'
    elif pd.notna(row['_lacoste_imaging_hit']):
        return 'Lacoste'
    return np.nan

full_df['imaging_hit_unified_source'] = full_df.apply(get_imaging_source, axis=1)

# Flag discordance
def get_imaging_discordance(row):
    if pd.notna(row['imaging_hit']) and pd.notna(row['_lacoste_imaging_hit']):
        if row['imaging_hit'] != row['_lacoste_imaging_hit']:
            return True
        else:
            return False
    return np.nan

full_df['imaging_hit_discordant'] = full_df.apply(get_imaging_discordance, axis=1)

# Clean up temp column
full_df.drop(columns=['_lacoste_imaging_hit'], inplace=True)

print(f"\n=== imaging_hit_unified ===")
print(full_df['imaging_hit_unified'].value_counts(dropna=False))
print(f"\n  Source breakdown:")
print(full_df['imaging_hit_unified_source'].value_counts(dropna=False))
print(f"\n  Discordant flags:")
print(full_df['imaging_hit_discordant'].value_counts(dropna=False))


# -------------------------------------------------------
# 2. ABUNDANCE HIT UNIFIED
# -------------------------------------------------------

print("\n\n" + "=" * 60)
print("=== Abundance columns audit ===")
print(f"\ndualipa_hit (1pct):")
print(full_df['dualipa_hit'].value_counts(dropna=False))
print(f"\nSahni_Chaperone_binding:")
print(full_df['Sahni_Chaperone_binding'].value_counts(dropna=False))
print(f"\nFragoza_Abundance_Status:")
print(full_df['Fragoza_Abundance_Status'].value_counts(dropna=False))

# Harmonize Sahni_Chaperone_binding: Y (binds chaperone = unstable) -> True, N -> False
full_df['_sahni_abun_hit'] = full_df['Sahni_Chaperone_binding'].map({'Y': True, 'N': False})

# Harmonize Fragoza_Abundance_Status: unstable/partially_unstable -> True, stable -> False
full_df['_fragoza_abun_hit'] = full_df['Fragoza_Abundance_Status'].map({
    'unstable': True,
    'partially_unstable': True,
    'stable': False,
})

# Check pairwise concordance
abun_sources = {
    '1pct': 'dualipa_hit',
    'Sahni': '_sahni_abun_hit',
    'Fragoza': '_fragoza_abun_hit',
}

print(f"\n=== Abundance concordance (pairwise) ===")
for (name_a, col_a), (name_b, col_b) in combinations(abun_sources.items(), 2):
    both = full_df[full_df[col_a].notna() & full_df[col_b].notna()].copy()
    if len(both) == 0:
        print(f"\n{name_a} ∩ {name_b}: no overlap")
        continue
    
    both['_concordant'] = both[col_a] == both[col_b]
    concordant = both['_concordant'].sum()
    discordant = (~both['_concordant']).sum()
    
    print(f"\n{name_a} ∩ {name_b}: {len(both)} overlapping variants")
    print(f"  Concordant: {concordant} ({100*concordant/len(both):.1f}%)")
    print(f"  Discordant: {discordant} ({100*discordant/len(both):.1f}%)")
    
    if discordant > 0:
        ct = pd.crosstab(
            both[col_a].map({True: f'{name_a}_True', False: f'{name_a}_False'}),
            both[col_b].map({True: f'{name_b}_True', False: f'{name_b}_False'}),
            margins=True
        )
        print(f"  Cross-tab:")
        print(ct.to_string())
        
        disc = both[~both['_concordant']]
        print(f"\n  Discordant variants:")
        show_cols = ['gene_variant', col_a, col_b, 'Source_Simple']
        # Also show original columns for context
        extra_cols = ['Sahni_Chaperone_binding', 'Fragoza_Abundance_Status', 'dualipa_hit']
        show_cols += [c for c in extra_cols if c not in show_cols and c in full_df.columns]
        print(disc[show_cols].to_string(index=False))

# Build unified column: 1pct > Sahni > Fragoza
full_df['abundance_hit_unified'] = (
    full_df['dualipa_hit']
    .where(full_df['dualipa_hit'].notna(), full_df['_sahni_abun_hit'])
    .where(full_df['dualipa_hit'].notna() | full_df['_sahni_abun_hit'].notna(), full_df['_fragoza_abun_hit'])
)

# Track source
def get_abundance_source(row):
    if pd.notna(row['dualipa_hit']):
        return '1pct'
    elif pd.notna(row['_sahni_abun_hit']):
        return 'Sahni'
    elif pd.notna(row['_fragoza_abun_hit']):
        return 'Fragoza'
    return np.nan

full_df['abundance_hit_unified_source'] = full_df.apply(get_abundance_source, axis=1)

# Flag discordance: check if ANY pair of available sources disagree
def get_abundance_discordance(row):
    values = {}
    if pd.notna(row['dualipa_hit']):
        values['1pct'] = row['dualipa_hit']
    if pd.notna(row['_sahni_abun_hit']):
        values['Sahni'] = row['_sahni_abun_hit']
    if pd.notna(row['_fragoza_abun_hit']):
        values['Fragoza'] = row['_fragoza_abun_hit']
    
    if len(values) < 2:
        return np.nan  # Can't assess discordance with <2 sources
    
    unique_vals = set(values.values())
    return len(unique_vals) > 1  # True if discordant

full_df['abundance_hit_discordant'] = full_df.apply(get_abundance_discordance, axis=1)

# Clean up temp columns
full_df.drop(columns=['_sahni_abun_hit', '_fragoza_abun_hit'], inplace=True)

print(f"\n=== abundance_hit_unified ===")
print(full_df['abundance_hit_unified'].value_counts(dropna=False))
print(f"\n  Source breakdown:")
print(full_df['abundance_hit_unified_source'].value_counts(dropna=False))
print(f"\n  Discordant flags:")
print(full_df['abundance_hit_discordant'].value_counts(dropna=False))

# -------------------------------------------------------
# 3. SUMMARY
# -------------------------------------------------------
print("\n\n" + "=" * 60)
print("=== UNIFIED COLUMNS SUMMARY ===")
print(f"\nTotal variants: {len(full_df)}")
print(f"  With imaging_hit_unified:   {full_df['imaging_hit_unified'].notna().sum()}")
print(f"    of which discordant:      {(full_df['imaging_hit_discordant'] == True).sum()}")
print(f"  With abundance_hit_unified: {full_df['abundance_hit_unified'].notna().sum()}")
print(f"    of which discordant:      {(full_df['abundance_hit_discordant'] == True).sum()}")
```

```python
# Step 1: First merge on current gene_variant (as before)
existing_cols = set(full_df.columns)
allele_col_new_cols = [c for c in allele_col_dedup.columns if c not in existing_cols or c == 'gene_variant']

print(f"Columns from allele_col to add: {len(allele_col_new_cols) - 1}")
print(allele_col_new_cols)

full_df_annotated = full_df.merge(
    allele_col_dedup[allele_col_new_cols],
    on='gene_variant',
    how='left',
)

n_missing_before = full_df_annotated[allele_col_new_cols[2]].isna().sum()  # pick first non-key col
print(f"\nAfter primary merge: {n_missing_before} variants without allele_col annotation")

# Step 2: For rows still missing annotations, try matching via old gene symbol
missing_mask = full_df_annotated[allele_col_new_cols[2]].isna()
has_old_symbol = full_df_annotated['gene_symbol_old'].notna()
backfill_mask = missing_mask & has_old_symbol

print(f"Variants to attempt backfill with gene_symbol_old: {backfill_mask.sum()}")

if backfill_mask.sum() > 0:
    # Build temp gene_variant using old symbol
    full_df_annotated['_gene_variant_old'] = (
        full_df_annotated['gene_symbol_old'].str.split('; ').str[0] + '_' + full_df_annotated['aa_change']
    )
    
    # Merge allele_col on old gene_variant
    allele_col_for_backfill = allele_col_dedup[allele_col_new_cols].rename(
        columns={'gene_variant': '_gene_variant_old'}
    )
    
    backfill_df = full_df_annotated.loc[backfill_mask, ['_gene_variant_old']].merge(
        allele_col_for_backfill,
        on='_gene_variant_old',
        how='left',
    )
    
    # Fill in the missing values
    n_backfilled = 0
    for col in allele_col_new_cols:
        if col == 'gene_variant':
            continue
        if col in backfill_df.columns:
            new_vals = backfill_df[col].values
            old_vals = full_df_annotated.loc[backfill_mask, col].values
            fill_mask = pd.isna(old_vals) & pd.notna(new_vals)
            if fill_mask.any():
                full_df_annotated.loc[backfill_mask, col] = np.where(
                    pd.isna(full_df_annotated.loc[backfill_mask, col]),
                    backfill_df[col].values,
                    full_df_annotated.loc[backfill_mask, col]
                )
                n_backfilled += fill_mask.sum()
    
    # Clean up temp column
    full_df_annotated.drop(columns=['_gene_variant_old'], inplace=True)
    
    n_missing_after = full_df_annotated[allele_col_new_cols[2]].isna().sum()
    n_rescued = n_missing_before - n_missing_after
    print(f"Rescued {n_rescued} variants via gene_symbol_old backfill")
    print(f"Still missing: {n_missing_after}")

# Final check
print(f"\nfull_df: {full_df.shape} → full_df_annotated: {full_df_annotated.shape}")
assert len(full_df_annotated) == len(full_df), f"Row count changed!"

# Show remaining unannotated variants by source
still_missing = full_df_annotated[full_df_annotated[allele_col_new_cols[2]].isna()]
print(f"\nRemaining unannotated by Source_Simple:")
print(still_missing['Source_Simple'].value_counts(dropna=False))

full_df_annotated
```
