# Sequencing QC Report

Generate summary statistics for why MVK variants were dropped during sequencing confirmation across different assay formats.

```python
# Define assay columns to analyze
assay_cols = {
    'entry': ['entry_sequenced', 'entry_sequence_confirmation_class', 'entry_seq_final'],
    'db': ['db_sequenced', 'db_sequence_confirmation_class', 'db_seq_final'],
    'dualip': ['dualip_sequenced', 'dualip_sequence_confirmation_class', 'dualip_seq_final'],
    'mislocalization': ['mislocalization_sequenced', 'mislocalization_sequence_confirmation_class', 'mislocalization_seq_final']
}

# Generate report for each assay
reports = {}

for assay_name, cols in assay_cols.items():
    seq_col, class_col, final_col = cols
    
    # Check if columns exist in the dataframe
    if not all(col in seq_df.columns for col in cols):
        print(f"Skipping {assay_name}: columns not found")
        continue
    
    # Filter to variants only (mut_id > 0)
    assay_df = seq_df[seq_df.mut_id > 0].copy()
    
    # Overall stats
    total_variants = len(assay_df)
    sequenced = assay_df[seq_col].notna().sum()
    sequenced_pct = (sequenced / total_variants * 100) if total_variants > 0 else 0
    
    # Final call breakdown
    final_Y = (assay_df[final_col] == 'Y').sum()
    final_N = (assay_df[final_col] == 'N').sum()
    
    # Sequence confirmation class breakdown (for sequenced variants)
    seq_variants = assay_df[assay_df[seq_col].notna()]
    class_counts = seq_variants[class_col].value_counts().to_dict()
    
    # Reasons for final=False (N or M)
    failed = assay_df[assay_df[final_col].isin(['N', 'M'])]
    failed_class_counts = failed[class_col].value_counts().to_dict()
    
    reports[assay_name] = {
        'total_variants': total_variants,
        'sequenced': sequenced,
        'sequenced_pct': sequenced_pct,
        'final_Y': final_Y,
        'final_N': final_N,
        'pass_rate': (final_Y / sequenced * 100) if sequenced > 0 else 0,
        'all_class_counts': class_counts,
        'failed_class_counts': failed_class_counts
    }

# Print formatted reports
for assay_name, report in reports.items():
    print(f"\n{'='*60}")
    print(f"SEQUENCING QC REPORT: {assay_name.upper()}")
    print(f"{'='*60}")
    print(f"\nOverall Statistics:")
    print(f"  Total MVK variants: {report['total_variants']}")
    print(f"  Sequenced: {report['sequenced']} ({report['sequenced_pct']:.1f}%)")
    print(f"\nFinal Call Breakdown:")
    print(f"  Y (Pass): {report['final_Y']} ({report['pass_rate']:.1f}% of sequenced)")
    print(f"  N (Fail): {report['final_N']}")
    print(f"\nSequence Confirmation Classes (All Sequenced Variants):")
    for class_type, count in sorted(report['all_class_counts'].items()):
        print(f"  {class_type}: {count}")
    print(f"\nReasons for Failure (N or M only):")
    if report['failed_class_counts']:
        for class_type, count in sorted(report['failed_class_counts'].items()):
            print(f"  {class_type}: {count}")
    else:
        print("  All variants passed!")

print(f"\n{'='*60}\n")
```

## Detailed Failure Analysis

List specific variants that failed sequencing QC for each assay.

```python
# Show specific variants that failed for each assay
display_cols = ['aa_change', 'mut_id']

for assay_name, cols in assay_cols.items():
    seq_col, class_col, final_col = cols
    
    if not all(col in seq_df.columns for col in cols):
        continue
    
    print(f"\n{'='*60}")
    print(f"FAILED VARIANTS: {assay_name.upper()}")
    print(f"{'='*60}\n")
    
    # Get failed variants
    failed = seq_df[(seq_df.mut_id > 0) & (seq_df[final_col].isin(['N', 'M']))].copy()
    
    if len(failed) == 0:
        print("No variants failed!\n")
        continue
    
    # Display by failure reason
    for class_type in failed[class_col].unique():
        if pd.notna(class_type):
            variants_in_class = failed[failed[class_col] == class_type]
            print(f"{class_type} ({len(variants_in_class)} variants):")
            for _, row in variants_in_class.iterrows():
                print(f"  - {row['aa_change']} (mut_id: {row['mut_id']}, final: {row[final_col]})")
            print()
```

## Global Sequencing QC Report

Variants must pass sequencing confirmation in **all 4 assays** (entry, db, dualipa, mislocalization) to be considered validated.

```python
# Global sequencing validation requiring all 4 assays to pass
variants_df = seq_df[seq_df.mut_id > 0].copy()

# Check if all 4 seq_final columns are 'Y'
all_assays = ['entry_seq_final', 'db_seq_final', 'dualip_seq_final', 'mislocalization_seq_final']

# Create a pass/fail for each assay
for assay in all_assays:
    variants_df[f'{assay}_pass'] = variants_df[assay] == 'Y'

# Global validation requires ALL 4 to be 'Y'
variants_df['global_seq_validated'] = (
    (variants_df['entry_seq_final'] == 'Y') &
    (variants_df['db_seq_final'] == 'Y') &
    (variants_df['dualip_seq_final'] == 'Y') &
    (variants_df['mislocalization_seq_final'] == 'Y')
)

# Summary statistics
total_variants = len(variants_df)
globally_validated = variants_df['global_seq_validated'].sum()
validation_rate = (globally_validated / total_variants * 100) if total_variants > 0 else 0

print(f"\n{'='*60}")
print(f"GLOBAL SEQUENCING QC REPORT (ALL 4 ASSAYS MUST PASS)")
print(f"{'='*60}\n")
print(f"Total MVK variants: {total_variants}")
print(f"Globally validated (all 4 assays pass): {globally_validated} ({validation_rate:.1f}%)")
print(f"Failed at least one assay: {total_variants - globally_validated} ({100-validation_rate:.1f}%)\n")

# Breakdown by number of assays passed
pass_counts = variants_df[['entry_seq_final_pass', 'db_seq_final_pass', 
                            'dualip_seq_final_pass', 'mislocalization_seq_final_pass']].sum(axis=1)
print("Distribution by number of assays passed:")
for i in range(5):
    count = (pass_counts == i).sum()
    pct = (count / total_variants * 100) if total_variants > 0 else 0
    print(f"  {i}/4 assays passed: {count} ({pct:.1f}%)")

print(f"\n{'='*60}\n")

# Show which specific assays failed for variants that didn't pass all 4
failed_global = variants_df[~variants_df['global_seq_validated']].copy()

if len(failed_global) > 0:
    print(f"\nVariants failing at least one assay ({len(failed_global)} total):\n")
    
    for _, row in failed_global.iterrows():
        failed_assays = []
        for assay in ['entry', 'db', 'dualip', 'mislocalization']:
            if row[f'{assay}_seq_final'] != 'Y':
                failed_assays.append(f"{assay} ({row[f'{assay}_seq_final']})")
        
        print(f"  {row['aa_change']} (mut_id: {row['mut_id']}):")
        print(f"    Failed: {', '.join(failed_assays)}")
else:
    print("\nAll variants passed all 4 assays!")
```

```python
# Global sequencing validation requiring all 4 assays to pass
variants_df = seq_df[seq_df.mut_id > 0].copy()

# Check if all 4 seq_final columns are 'Y'
all_assays = ['entry_seq_final', 'db_seq_final', 'dualip_seq_final', 'mislocalization_seq_final']

# Global validation requires ALL 4 to be 'Y'
variants_df['global_seq_validated'] = (
    (variants_df['entry_seq_final'] == 'Y') &
    (variants_df['db_seq_final'] == 'Y') &
    (variants_df['dualip_seq_final'] == 'Y') &
    (variants_df['mislocalization_seq_final'] == 'Y')
)

# Create comprehensive table
table_data = []

for _, row in variants_df.iterrows():
    # Get failure information for each assay
    entry_status = row['entry_seq_final']
    entry_class = row['entry_sequence_confirmation_class'] if entry_status != 'Y' else '-'
    
    db_status = row['db_seq_final']
    db_class = row['db_sequence_confirmation_class'] if db_status != 'Y' else '-'
    
    dualip_status = row['dualip_seq_final']
    dualip_class = row['dualip_sequence_confirmation_class'] if dualip_status != 'Y' else '-'
    
    misloc_status = row['mislocalization_seq_final']
    misloc_class = row['mislocalization_sequence_confirmation_class'] if misloc_status != 'Y' else '-'
    
    table_data.append({
        'Variant': row['aa_change'],
        'mut_id': row['mut_id'],
        'Global Status': 'PASS' if row['global_seq_validated'] else 'FAIL',
        'Entry': entry_status,
        'Entry Class': entry_class,
        'DB': db_status,
        'DB Class': db_class,
        'DualIP': dualip_status,
        'DualIP Class': dualip_class,
        'Misloc': misloc_status,
        'Misloc Class': misloc_class
    })

# Create DataFrame
result_df = pd.DataFrame(table_data)

# Sort: PASS first, then FAIL
result_df = result_df.sort_values(by=['Global Status', 'Variant'], ascending=[False, True])

# Print summary
total_variants = len(result_df)
passed = (result_df['Global Status'] == 'PASS').sum()
failed = (result_df['Global Status'] == 'FAIL').sum()

print(f"\n{'='*100}")
print(f"COMPLETE MVK SEQUENCING QC TABLE")
print(f"{'='*100}\n")
print(f"Total variants: {total_variants}")
print(f"Passed all 4 assays: {passed} ({passed/total_variants*100:.1f}%)")
print(f"Failed at least one assay: {failed} ({failed/total_variants*100:.1f}%)\n")
print(f"{'='*100}\n")

# Display the table
print(result_df.to_string(index=False))

print(f"\n{'='*100}")
print("\nLEGEND:")
print("Status: Y = Pass, N = Fail, M = Manual review needed")
print("\nSequence Confirmation Classes:")
print("  1 = perfectly validated")
print("  2 = partially validated, >= 50% coverage")
print("  3 = wild type")
print("  4 = partial wild type, >= 50% coverage")
print("  5 = target and off-target mutation, >= 50% coverage")
print("  6 = off-target mutation, >= 50% coverage")
print("  7 = truncated, < 50% coverage")
print("  99 = no reads")
print(f"{'='*100}\n")

# Optionally save to file
# result_df.to_csv(output_folder / 'MVK_sequencing_QC_complete.tsv', sep='\t', index=False)
# print(f"Table saved to: {output_folder / 'MVK_sequencing_QC_complete.tsv'}")
```
