```python
# no MLH1 interactions with confident AF3 structures, lets look for any AF3 structures with MLH1 and check confidence scores

def load_probable_residue_contacts(in_dir, cutoff=0.95):
    in_dir = Path(in_dir)
    rrc = []
    for fpath in in_dir.glob("*/*_confidences.json"):
        if fpath.stem.endswith("summary_confidences"):
            continue
        pair = fpath.parents[0].stem
        len_a, len_b = get_lengths_from_PDB_file(_find_pdb_file_path(fpath.parent))
        with open(fpath, "r") as f:
            conf = json.load(f)
        cp = np.array(conf["contact_probs"])
        pairs = pd.DataFrame(
            data=np.array(np.where(cp > cutoff)).T,
            columns=["Residue_index_X", "Residue_index_Y"],
        )
        pairs["pair"] = pair
        pairs["contact_probability"] = pairs.apply(
            lambda row: cp[row["Residue_index_X"], row["Residue_index_Y"]], axis=1
        )
        pairs = pairs.loc[
            (pairs["Residue_index_X"] < len_a) & (pairs["Residue_index_Y"] >= len_a), :
        ]
        pairs["Residue_index_X"] = pairs["Residue_index_X"] + 1
        pairs["Residue_index_Y"] = (pairs["Residue_index_Y"] - len_a) + 1
        rrc.append(pairs.copy())
    rrc = pd.concat(rrc)
    rrc["gene_name_a"] = rrc["pair"].apply(lambda x: x.split("_")[0])
    rrc["gene_name_b"] = rrc["pair"].apply(lambda x: x.split("_")[1])
    return rrc
```

```python
in_dir = '/Users/user_42/Partners HealthCare Dropbox/Maxime Tixhon/VarChamp/predicted_structures'
gene_filter = 'p40692'  # MLH1 UniProt ID
cutoff = 0.1

in_dir = Path(in_dir)
rrc = []

for fpath in in_dir.glob("*/*_confidences.json"):
    if fpath.stem.endswith("summary_confidences"):
        continue
    
    pair = fpath.parents[0].stem
    
    # Filter for MLH1-containing pairs
    if gene_filter.lower() not in pair.lower():
        continue
    
    len_a, len_b = get_lengths_from_PDB_file(_find_pdb_file_path(fpath.parent))
    
    with open(fpath, "r") as f:
        conf = json.load(f)
    
    cp = np.array(conf["contact_probs"])
    
    pairs = pd.DataFrame(
        data=np.array(np.where(cp > cutoff)).T,
        columns=["Residue_index_X", "Residue_index_Y"],
    )
    
    
    
    if len(pairs) == 0:
        continue
        
    pairs["pair"] = pair
    pairs["contact_probability"] = pairs.apply(
        lambda row: cp[row["Residue_index_X"], row["Residue_index_Y"]], axis=1
    )
    
    
    # Filter for inter-chain contacts only
    pairs = pairs.loc[
        (pairs["Residue_index_X"] < len_a) & (pairs["Residue_index_Y"] >= len_a), :
    ]
    
    if len(pairs) == 0:
        continue
    

    
    # Adjust indices to 1-based
    pairs["Residue_index_X"] = pairs["Residue_index_X"] + 1
    pairs["Residue_index_Y"] = (pairs["Residue_index_Y"] - len_a) + 1
    
    rrc.append(pairs.copy())

if len(rrc) == 0:
    print(f"No contacts found for {gene_filter} above cutoff {cutoff}")
    # return pd.DataFrame()

rrc = pd.concat(rrc)
rrc["gene_name_a"] = rrc["pair"].apply(lambda x: x.split("_")[0])
rrc["gene_name_b"] = rrc["pair"].apply(lambda x: x.split("_")[1])


rrc
# return rrc

# # # Usage:
# # mlh1_contacts = load_probable_residue_contacts_mlh1(
# # '/Users/user_42/Partners HealthCare Dropbox/Maxime Tixhon/VarChamp/predicted_structures',
# # cutoff=0.95,
# # gene_filter='p40692'  # MLH1 UniProt ID
# # )

# # print(f"Found {len(mlh1_contacts)} contacts for MLH1")
# # print(f"Number of unique MLH1 interaction pairs: {mlh1_contacts['pair'].nunique()}")
# # print("\nMLH1 interaction partners:")
# # print(mlh1_contacts.groupby('pair').size().sort_values(ascending=False))
```
